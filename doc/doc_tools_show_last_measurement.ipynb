# WATCHDOG – Erklärung von show_last_measurements.py

Dieses Notebook dokumentiert die Logik des Tools `show_last_measurements.py`.

## Zweck des Skripts

Das Tool dient als Diagnose- und Servicewerkzeug. Es prüft zunächst, ob die SQLite-Datenbank des WATCHDOG-Systems vorhanden ist. Erst danach können Messwerte ausgelesen und angezeigt werden.

## Originalcode
```python
import sqlite3
from pathlib import Path

DATABASE_PATH = Path("../data/watchdog.db")

def main():
    if not DATABASE_PATH.exists():
        print("Database not found - you may want to create one.")
        return

if __name__ == "__main__":
    main()
```

## Importe

### sqlite3
Ermöglicht den Zugriff auf die lokale SQLite-Datenbank des WATCHDOG-Systems.

### Path
`pathlib.Path` wird verwendet, um Dateipfade plattformunabhängig zu verwalten.

## Datenbankpfad

```python
DATABASE_PATH = Path("../data/watchdog.db")
```

Der Pfad verweist auf die zentrale Datenbank:

```text
Watchdog/
├── data/
│   └── watchdog.db
└── tools/
    └── show_last_measurements.py
```

Durch `../` wird vom Ordner `tools/` eine Ebene nach oben gewechselt.

## Ablauf der Funktion main()

### Schritt 1
Prüfen, ob die Datenbank existiert.

```python
if not DATABASE_PATH.exists():
```

### Schritt 2
Falls keine Datenbank gefunden wird:

```python
print("Database not found - you may want to create one.")
return
```

Das Skript beendet sich kontrolliert, anstatt einen Fehler auszulösen.

## Programmstart

```python
if __name__ == "__main__":
    main()
```

Damit wird `main()` nur ausgeführt, wenn das Skript direkt gestartet wird.

## Geplanter Ausbau

Das aktuelle Skript scheint noch ein Grundgerüst zu sein. Typischer nächster Schritt:

```python
conn = sqlite3.connect(DATABASE_PATH)
cursor = conn.cursor()

cursor.execute("""
SELECT timestamp, measurement_name, value
FROM measurements
ORDER BY timestamp DESC
LIMIT 20
""")
```

Danach könnten die letzten Messwerte ausgegeben werden.
