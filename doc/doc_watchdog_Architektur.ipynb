# WATCHDOG – Architektur-Dokumentation

Dieses Notebook beschreibt die Architektur des WATCHDOG-Projekts.

## Projektziel
- Modbus-Datenerfassung
- Speicherung historischer Messwerte
- Analyse und Fernüberwachung

## Projektstruktur
```text
watchdog/
├── config.py
├── database.py
├── logger.py
├── main.py
├── modbus_client.py
└── register_map.py
```

## Gesamtarchitektur
```text
Modbus Gerät → modbus_client.py → register_map.py → main.py → database.py
                                                     ↘ logger.py
```

## Komponenten
### config.py
Zentrale Konfiguration

### modbus_client.py
Kommunikation mit der Anlage

### database.py
Speicherung der Messwerte

### logger.py
Protokollierung

### register_map.py
Zuordnung der Register zu Messwerten

## Roadmap
1. Register Map vervollständigen
2. Alarmmanagement ergänzen
3. Kommunikationsüberwachung
4. Dashboard entwickeln
5. Fernwartung erweitern
