# WATCHDOG – Erklärung von logger.py

Dieses Notebook beschreibt die Logik der Datei `watchdog/logger.py` und dient als Entwicklerdokumentation für VS Code Jupyter.

## Zweck des Moduls
Das Modul stellt die zentrale Logging-Konfiguration für WATCHDOG bereit.

Ablauf:
1. Konfiguration aus `APP_CONFIG` lesen
2. Pfad zur Logdatei bestimmen
3. Log-Verzeichnis automatisch anlegen
4. Python-Logging konfigurieren
5. Meldungen in Datei und optional auf Konsole ausgeben

## Originalcode
```python
import logging
from pathlib import Path
from watchdog.config import APP_CONFIG

def setup_logger():
    log_file = Path(APP_CONFIG["log_file"])
    log_file.parent.mkdir(exist_ok=True)
```

## Detaillierte Erklärung

### Import von logging
Das Standardmodul `logging` wird verwendet, um Statusmeldungen, Warnungen und Fehler zu protokollieren.

### Import von Path
`pathlib.Path` ermöglicht plattformunabhängige Arbeit mit Dateipfaden.

### APP_CONFIG
Die Konfiguration enthält unter anderem den Pfad zur Logdatei, z.B.:
```python
APP_CONFIG = {
    "log_file": "logs/watchdog.log"
}
```

### setup_logger()
Diese Funktion bereitet das Logging-System vor.

```python
log_file = Path(APP_CONFIG["log_file"])
```
Erzeugt ein Path-Objekt für die konfigurierte Logdatei.

```python
log_file.parent.mkdir(exist_ok=True)
```
Legt den Ordner `logs/` automatisch an, falls er noch nicht existiert.

## Geplanter vollständiger Ablauf
```text
run_watchdog.py
      ↓
setup_logger()
      ↓
logs/ wird geprüft
      ↓
watchdog.log wird geöffnet
      ↓
Messungen, Fehler und Statusmeldungen werden gespeichert
```

## Verbesserungsvorschlag

```python
logging.basicConfig(
    filename=log_file,
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(message)s"
)
```

Damit würden alle WATCHDOG-Komponenten zentral protokolliert werden.
