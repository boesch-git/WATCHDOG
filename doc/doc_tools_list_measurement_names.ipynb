# WATCHDOG – Erklärung von list_measurement_names.py

Dieses Notebook dokumentiert die Logik des Tools `list_measurement_names.py`.

## Zweck des Skripts

Das Tool soll prüfen, ob die WATCHDOG-Datenbank vorhanden ist. Später kann es erweitert werden, um alle verfügbaren Messnamen aus der SQLite-Datenbank aufzulisten.

## Originalcode
```python
import sqlite3
from pathlib import Path

DATABASE_PATH = Path("data/watchdog.db")

def main():
    if not DATABASE_PATH.exists():
        print("Datenbank existiert nicht!", DATABASE_PATH)
        return

if __name__ == "__main__":
    main()
```

## Importe

### sqlite3
Ermöglicht den Zugriff auf die SQLite-Datenbank des WATCHDOG-Systems.

### pathlib.Path
Wird verwendet, um Dateipfade plattformunabhängig zu verwalten.

## Datenbankpfad

```python
DATABASE_PATH = Path("data/watchdog.db")
```

Der Kommentar im Code weist darauf hin, dass noch geklärt werden muss, ob relative oder absolute Pfade verwendet werden sollen.

## Ablauf der Funktion main()

1. Prüfung, ob die Datenbank existiert.
2. Ausgabe einer Fehlermeldung, falls die Datei fehlt.
3. Kontrolliertes Beenden des Programms.

## Einstiegspunkt

```python
if __name__ == "__main__":
    main()
```

Dadurch wird das Skript nur bei direktem Start ausgeführt.

## Geplanter Ausbau

Später können alle vorhandenen Messgrößen mit einer SQL-Abfrage ermittelt und ausgegeben werden.
