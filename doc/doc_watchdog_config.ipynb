# WATCHDOG Konfigurationsdokumentation

Dieses Notebook dokumentiert die Datei `watchdog/config.py`.

## Zweck
Die Konfiguration wird aktuell direkt im Python-Code gehalten. Später ist eine Migration auf `watchdog.yaml` vorgesehen.

```python
MODBUS_CONFIG = {
    "port": "/dev/ttyUSB0",
    "baudrate": 9600,
    "bytesize": 8,
    "parity": "N",
    "stopbits": 1,
    "timeout": 2,
    "slave_id": 1,
    "poll_interval_seconds": 5,
}

APP_CONFIG = {
    "log_file": "logs/watchdog.log",
}

DATABASE_CONFIG = {
    "path": "data/watchdog.db",
}
```

## MODBUS_CONFIG
Konfiguration der Modbus-RTU-Kommunikation.

- **port**: Serieller Port
- **baudrate**: Übertragungsgeschwindigkeit
- **bytesize**: Anzahl Datenbits
- **parity**: Parität (N/E/O)
- **stopbits**: Stopbits
- **timeout**: Timeout in Sekunden
- **slave_id**: Modbus-Adresse
- **poll_interval_seconds**: Abfrageintervall

## APP_CONFIG
Anwendungseinstellungen.

- **log_file**: Speicherort der Logdatei

## DATABASE_CONFIG
Lokale SQLite-Datenbank.

- **path**: Pfad zur Datenbankdatei

## ADR-001
### Entscheidung
Die Konfiguration wird zunächst in `config.py` über Python-Dictionaries verwaltet.

### Vorteile
- Einfach
- Schnell implementiert
- Keine zusätzlichen Abhängigkeiten

### Nachteile
- Keine Trennung von Code und Konfiguration
- Keine Umgebungsprofile
- Änderungen erfordern Neustart

### Zukunft
Migration auf `watchdog.yaml` mit Validierung und Profilen.
