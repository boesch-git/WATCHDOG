## Verbindungsaufbau

Beim Verbindungsaufbau werden folgende SQLite-Optimierungen aktiviert:

```sql
PRAGMA journal_mode=WAL;
PRAGMA synchronous=NORMAL;
```

**WAL-Modus:** bessere Parallelität zwischen Lesen und Schreiben.
**NORMAL:** guter Kompromiss zwischen Performance und Datensicherheit.

## Messwertspeicherung

Die Methode `insert_measurement()` erstellt automatisch einen UTC-Zeitstempel und speichert den Datensatz mittels parametrisierter SQL-Statements.

Vorteile:
- Schutz vor SQL-Injection
- Konsistente Zeitbasis
- Flexible Erweiterung des Messwertobjekts


## ADR-002: SQLite als lokale Persistenz

### Status
Accepted

### Entscheidung
WATCHDOG verwendet SQLite als lokale Datenbank.

### Gründe
- Keine zusätzliche Serverinstallation
- Ideal für Raspberry Pi und Embedded-Systeme
- Einfache Datensicherung
- Wenig Wartungsaufwand

### Risiken
- Begrenzte Skalierbarkeit
- Keine verteilte Datenhaltung

### Zukunft
- Historisierung erweitern
- Indizes ergänzen
- Export zu Cloud- oder Timeseries-Datenbanken ermöglichen
